```python
import sys
import os
sys.path.append(os.path.abspath('..'))
import numpy as np
import pandas as pd
from data.synthetic_generator import generate_synthetic_activity_dataset
from algorithms.clustering import run_kmeans_clustering
from algorithms.anomaly import run_isolation_forest_anomaly_detection
from algorithms.association import run_apriori_association_mining
from algorithms.correlation import run_correlation_analysis
from algorithms.pacing import calculate_goal_progress_and_pacing
from algorithms.recommendations import generate_workflow_recommendations
from models.schemas import DailyRecord, GoalProgressRequest, MilestoneItem

print('All ML modules loaded successfully.')
```

## 1. Dataset Generation & Exploration
Generate 60 days of realistic student activity with habits, focus sessions, screen times, and milestone achievements.

```python
raw_data = generate_synthetic_activity_dataset(days_count=60)
df = pd.DataFrame(raw_data)
records = [DailyRecord(**r) for r in raw_data]
print(f'Total days: {len(df)}')
df[['date', 'focused_work_minutes', 'distracting_screen_time_minutes', 'habit_adherence_rate', 'actual_goal_progress_delta']].head()
```

## 2. K-Means Clustering on Daily Behavioral Patterns
Behavioral features are standardized. Actual goal progress delta is strictly held out as an outcome evaluation metric.

```python
cluster_res = run_kmeans_clustering(records)
print(f'Optimal k: {cluster_res.optimal_k}')
print(f'Silhouette Score: {cluster_res.silhouette_score}')
for c in cluster_res.clusters:
    print(f'Cluster {c.cluster_id} [{c.cluster_name}]: {c.sample_count} days | Avg Progress Delta: +{c.avg_goal_progress_delta}%')
```

## 3. Isolation Forest Anomaly Detection
Flags atypical days and explains anomalies via multivariate z-scores.

```python
anomaly_res = run_isolation_forest_anomaly_detection(records, contamination=0.1)
print(f'Total anomalies flagged: {anomaly_res.anomaly_count}')
print(f'User marked rest days: {anomaly_res.user_marked_rest_days}')
flagged = [a for a in anomaly_res.anomalies if a.is_anomaly][:3]
for a in flagged:
    print(f'Date: {a.date} | Score: {a.anomaly_score} | Summary: {a.summary}')
```

## 4. Apriori Association Rule Mining
Evaluates support, confidence, and lift of activity combinations without claiming causal proof.

```python
assoc_res = run_apriori_association_mining(records, min_support=0.12, min_confidence=0.45, min_lift=1.05)
print(f'Rules discovered: {assoc_res.rules_count}')
for r in assoc_res.rules[:5]:
    print(f'Rule: {r.antecedents} => {r.consequents} | Support: {r.support} | Conf: {r.confidence} | Lift: {r.lift}')
```

## 5. Statistical Correlations & Evidence-Based Recommendations

```python
corr_res = run_correlation_analysis(records)
for cp in corr_res.correlations:
    if 'Daily Goal Progress' in cp.feature_b:
        print(f'{cp.feature_a}: Pearson r = {cp.pearson_r} (p={cp.pearson_p_value}), Spearman rho = {cp.spearman_rho}')

rec_res = generate_workflow_recommendations(records, goal_pacing_gap=-6.0, active_goal_title='Complete 200 DSA Problems')
print(f'\nGenerated {len(rec_res.recommendations)} Workflow Recommendations with actionable hypotheses.')
```
